```python
#handling missing values
```

```python
df_clean = df.copy()
```

```python
df_clean['Income'].median()
```

```python
df_clean['Income'] = df_clean['Income'].fillna(
    df_clean['Income'].median()
)
```

```python
df_clean['Income'].isnull().sum()
```

```python
Convert Dt_Customer into datetime
```

```python
df_clean['Dt_Customer'] = pd.to_datetime(
    df_clean['Dt_Customer'],
    format='%d-%m-%Y'
)
```

```python
df_clean['Dt_Customer'].dtype
```

```python
Investigate unusual birth years
```

```python
df_clean[
    df_clean['Year_Birth'] < 1940
][['ID', 'Year_Birth', 'Income']]
```

```python
Investigate Income outliers
```

```python
df_clean.nlargest(
    10, 'Income'
)[['ID', 'Year_Birth', 'Income']]
```

```python
print("Before removing outliers:", df_clean.shape)
```

```python
# Remove customers born before 1940
df_clean = df_clean[df_clean['Year_Birth'] >= 1940]

# Remove the extreme income record
df_clean = df_clean[df_clean['Income'] != 666666]
```

```python
# Reset index
df_clean = df_clean.reset_index(drop=True)
```

```python
# Check after removal
print("After removing outliers:", df_clean.shape)
```

```python
print("Missing values:")
print(df_clean.isnull().sum())
```

```python
print("Duplicate records:", df_clean.duplicated().sum())
```

```python
df_clean['Age'] = 2014 - df_clean['Year_Birth']
```
