```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings('ignore')
```

```python
#Age distribution
plt.figure(figsize=(10, 5))

sns.histplot(data=df, x='Age', bins=30, kde=True)

plt.title('Customer Age Distribution')
plt.xlabel('Age')
plt.ylabel('Number of Customers')

plt.show()
```

```python
#Education distribution
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x='Education',
              order=df['Education'].value_counts().index)

plt.title('Customer Education Distribution')
plt.xticks(rotation=30)

plt.show()
```

```python
#Marital status distribution
plt.figure(figsize=(9, 5))

sns.countplot(data=df, x='Marital_Status',
              order=df['Marital_Status'].value_counts().index)

plt.title('Customer Marital Status Distribution')
plt.xticks(rotation=30)

plt.show()
```

```python
#Income Analysis
plt.figure(figsize=(10, 5))

sns.histplot(data=df, x='Income', bins=40, kde=True)

plt.title('Customer Income Distribution')
plt.xlabel('Income')
plt.ylabel('Number of Customers')

plt.show()
```

```python
df['Income'].describe()
```

```python
#Customer Spending Analysis
spending_columns = [
    'MntWines',
    'MntFruits',
    'MntMeatProducts',
    'MntFishProducts',
    'MntSweetProducts',
    'MntGoldProds'
]

df['Total_Spending'] = df[spending_columns].sum(axis=1)

df[['ID', 'Total_Spending']].head()
```

```python
avg_spending = df[spending_columns].mean().sort_values(
    ascending=False
)

print(avg_spending)
```

```python
plt.figure(figsize=(10, 5))

sns.barplot(x=avg_spending.values,
            y=avg_spending.index)

plt.title('Average Spending by Product Category')
plt.xlabel('Average Spending')
plt.ylabel('Product Category')

plt.show()
```

```python
#Marketing Campaign Analysis
campaign_columns = [
    'AcceptedCmp1',
    'AcceptedCmp2',
    'AcceptedCmp3',
    'AcceptedCmp4',
    'AcceptedCmp5',
    'Response'
]

campaign_acceptance = df[campaign_columns].sum()

print(campaign_acceptance)
```

```python
plt.figure(figsize=(9, 5))

sns.barplot(x=campaign_acceptance.index,
            y=campaign_acceptance.values)

plt.title('Customer Campaign Acceptance')
plt.xlabel('Campaign')
plt.ylabel('Number of Accepted Customers')

plt.show()
```

```python
#Correlation Analysis
numeric_df = df.select_dtypes(include=np.number)

plt.figure(figsize=(16, 10))

sns.heatmap(
    numeric_df.corr(),
    cmap='coolwarm',
    annot=False
)

plt.title('Customer Data Correlation Heatmap')

plt.show()
```

```python
# Spending columns
spending_columns = [
    'MntWines',
    'MntFruits',
    'MntMeatProducts',
    'MntFishProducts',
    'MntSweetProducts',
    'MntGoldProds'
]

# Purchase columns
purchase_columns = [
    'NumWebPurchases',
    'NumCatalogPurchases',
    'NumStorePurchases',
    'NumDealsPurchases'
]

# Campaign columns
campaign_columns = [
    'AcceptedCmp1',
    'AcceptedCmp2',
    'AcceptedCmp3',
    'AcceptedCmp4',
    'AcceptedCmp5',
    'Response'
]
```

```python
# Demographic summary
print(df['Age'].describe())
print(df['Income'].describe())

# Spending summary
print(df[spending_columns].mean().sort_values(ascending=False))

# Purchase summary
print(df[purchase_columns].mean().sort_values(ascending=False))

# Campaign summary
print(df[campaign_columns].sum().sort_values(ascending=False))
```

```python
#feature selection
# Select relevant customer features

selected_features = [
    'Age',
    'Income',
    'Kidhome',
    'Teenhome',
    'Recency',
    'MntWines',
    'MntFruits',
    'MntMeatProducts',
    'MntFishProducts',
    'MntSweetProducts',
    'MntGoldProds',
    'NumDealsPurchases',
    'NumWebPurchases',
    'NumCatalogPurchases',
    'NumStorePurchases',
    'NumWebVisitsMonth'
]

# Create feature dataset
X = df[selected_features].copy()
print("Selected Features:", X.shape)
print("\nFeature Columns:")
print(X.columns.tolist())

X.head()
```
