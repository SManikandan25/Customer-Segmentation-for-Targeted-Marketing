```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv(r'C:\Users\smk25\OneDrive\Documents\Customer Segmentation for Targeted Marketing\dataset\cleaned_customer_data.csv')
```

```python
#feature engineering
#Feature 1 – Total Spending
spending_columns = [
    'MntWines',
    'MntFruits',
    'MntMeatProducts',
    'MntFishProducts',
    'MntSweetProducts',
    'MntGoldProds'
]

df['Total_Spending'] = df[spending_columns].sum(axis=1)

print(df[['ID', 'Total_Spending']].head())
```

```python
#Feature 2 – Total Purchases
purchase_columns = [
    'NumWebPurchases',
    'NumCatalogPurchases',
    'NumStorePurchases'
]

df['Total_Purchases'] = df[purchase_columns].sum(axis=1)

print(df[['ID', 'Total_Purchases']].head())
```

```python
#Feature 3 – Family Size
df['Family_Size'] = (
    df['Kidhome'] + df['Teenhome'] + 1
)

print(df[['ID', 'Kidhome', 'Teenhome', 'Family_Size']].head())
```

```python
# Convert customer enrollment date
df['Dt_Customer'] = pd.to_datetime(
    df['Dt_Customer'],
    format='%Y-%m-%d'
)
```

```python
#. Feature 4 – Customer Tenure
reference_date = df['Dt_Customer'].max() + pd.Timedelta(days=1)

df['Customer_Tenure'] = (
    reference_date - df['Dt_Customer']
).dt.days

print("Reference Date:", reference_date)

print(
    df[['ID', 'Dt_Customer', 'Customer_Tenure']].head()
)
```

```python
#Feature 5 – Wine Spending Ratio
df['Wine_Spending_Ratio'] = np.where(
    df['Total_Spending'] > 0,
    df['MntWines'] / df['Total_Spending'],
    0
)


print(
    df[['ID', 'MntWines', 'Total_Spending',
        'Wine_Spending_Ratio']].head()
)
```

```python
#feature scaling
model_features = [
    'Age',
    'Income',
    'Recency',
    'Family_Size',
    'Customer_Tenure',
    'MntWines',
    'MntFruits',
    'MntMeatProducts',
    'MntFishProducts',
    'MntSweetProducts',
    'MntGoldProds',
    'NumWebPurchases',
    'NumCatalogPurchases',
    'NumStorePurchases'
]

X_model = df[model_features].copy()

print("Shape:", X_model.shape)
print(X_model.head())
```

```python
#Apply StandardScaler
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X_model)

X_scaled = pd.DataFrame(
    X_scaled,
    columns=X_model.columns,
    index=X_model.index
)

print(X_scaled.head())
```

```python
print("Scaled Shape:", X_scaled.shape)

print("\nMean:")
print(X_scaled.mean().round(2))

print("\nStandard Deviation:")
print(X_scaled.std().round(2))

print("\nMinimum:")
print(X_scaled.min().round(2))

print("\nMaximum:")
print(X_scaled.max().round(2))
```

```python
#Step 1: Import
from sklearn.cluster import KMeans
```

```python
#Step 2: Build initial model
kmeans = KMeans(
    n_clusters=4,
    init='k-means++',
    n_init=10,
    random_state=42
)

#Step 3: Fit and predict
cluster_labels = kmeans.fit_predict(X_scaled)
```

```python
#Elbow Method
#Step 1 – Import libraries
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
#Step 2 – Create K value
inertia_values = []

K_range = range(1, 11)

#Step 3 – Build multiple K-Means models

for k in K_range:

    kmeans = KMeans(
        n_clusters=k,
        init='k-means++',
        n_init=10,
        random_state=42
    )

    kmeans.fit(X_scaled)

    inertia_values.append(kmeans.inertia_)

#Step 4 – Print inertia values

for k, inertia in zip(K_range, inertia_values):
    print(f"K = {k}, Inertia = {inertia:.2f}")
```

```python
#Step 5 – Plot Elbow Curve

plt.figure(figsize=(8, 5))

plt.plot(
    K_range,
    inertia_values,
    marker='o'
)

plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia')
plt.title('Elbow Method for Optimal K')

plt.xticks(K_range)
plt.grid(True)

plt.show()
```

```python
#Silhouette Score
#Step 1 – Import libraries
from sklearn.metrics import silhouette_score
```

```python
#Step 2 – Calculate Silhouette Score for K = 2 to 10
silhouette_values = []

K_range = range(2, 11)

for k in K_range:

    kmeans = KMeans(
        n_clusters=k,
        init='k-means++',
        n_init=10,
        random_state=42
    )

    labels = kmeans.fit_predict(X_scaled)

    score = silhouette_score(
        X_scaled,
        labels
    )

    silhouette_values.append(score)
```

```python
#Step 3 – Print scores
for k, score in zip(K_range, silhouette_values):
    print(f"K = {k}, Silhouette Score = {score:.4f}")
```

```python
#Step 4 – Plot Silhouette Score graph
plt.figure(figsize=(8, 5))

plt.plot(
    K_range,
    silhouette_values,
    marker='o'
)

plt.xlabel('Number of Clusters (K)')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Score for Optimal K')

plt.xticks(K_range)
plt.grid(True)

plt.show()
```
